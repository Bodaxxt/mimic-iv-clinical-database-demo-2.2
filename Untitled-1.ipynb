```python
# ==============================================================================
# CLINICAL COHORT EXTRACTION & FEATURE ENGINEERING PIPELINE
# Dataset: MIMIC-IV Clinical Database Demo (v2.2)
# Target Application: ICU Monitoring / Deterioration Prediction (XGBoost)
# ==============================================================================

import os
import warnings
import numpy as np
import pandas as pd

# Suppress downcasting and chained assignment warnings for clean notebook output
warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 1000)

# ------------------------------------------------------------------------------
# 1. PATH CONFIGURATION & SAFE LOADER HELPER
# ------------------------------------------------------------------------------
# Define base paths. Handles both compressed (.csv.gz) and raw (.csv) files.
DATA_DIR = "."  # Set to your MIMIC-IV root directory if different

def load_mimic_table(subfolder: str, table_name: str, usecols: list = None) -> pd.DataFrame:
    """
    Helper function to load MIMIC-IV CSV files supporting both .csv and .csv.gz.
    """
    gz_path = os.path.join(DATA_DIR, subfolder, f"{table_name}.csv.gz")
    csv_path = os.path.join(DATA_DIR, subfolder, f"{table_name}.csv")
    
    if os.path.exists(gz_path):
        target_path = gz_path
    elif os.path.exists(csv_path):
        target_path = csv_path
    else:
        raise FileNotFoundError(f"Neither {gz_path} nor {csv_path} was found.")
        
    print(f"Loading '{table_name}' from {target_path}...")
    return pd.read_csv(target_path, usecols=usecols, low_memory=False)


# ------------------------------------------------------------------------------
# 2. BASE COHORT EXTRACTION & DEMOGRAPHICS
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 1: Extracting Base ICU Cohort & Patient Demographics\n" + "="*70)

# A. Load ICU Stays
df_icustays = load_mimic_table("icu", "icustays", [
    "subject_id", "hadm_id", "stay_id", "first_careunit", "intime", "outtime", "los"
])
df_icustays["intime"] = pd.to_datetime(df_icustays["intime"])
df_icustays["outtime"] = pd.to_datetime(df_icustays["outtime"])

# B. Load Patients (anchor_age, anchor_year, gender)
df_patients = load_mimic_table("hosp", "patients", [
    "subject_id", "gender", "anchor_age", "anchor_year", "dod"
])

# C. Load Admissions (admittime, dischtime, mortality flag)
df_admissions = load_mimic_table("hosp", "admissions", [
    "subject_id", "hadm_id", "admittime", "dischtime", "admission_type", "hospital_expire_flag"
])
df_admissions["admittime"] = pd.to_datetime(df_admissions["admittime"])
df_admissions["dischtime"] = pd.to_datetime(df_admissions["dischtime"])

# D. Merge Demographics & Compute Exact Admission Age
# MIMIC-IV formula: age_at_admission = anchor_age + (admittime.year - anchor_year)
cohort_df = df_icustays.merge(df_patients, on="subject_id", how="inner")
cohort_df = cohort_df.merge(df_admissions, on=["subject_id", "hadm_id"], how="inner")

cohort_df["age_at_admission"] = cohort_df["anchor_age"] + (
    cohort_df["admittime"].dt.year - cohort_df["anchor_year"]
)
# Encode Gender: Female -> 1, Male -> 0
cohort_df["is_female"] = (cohort_df["gender"].str.upper() == "F").astype(int)

# Filter adult ICU stays (age >= 18) and positive stay lengths
cohort_df = cohort_df[cohort_df["age_at_admission"] >= 18].copy()
cohort_df = cohort_df[cohort_df["los"] > 0].copy()

base_cols = [
    "stay_id", "subject_id", "hadm_id", "intime", "outtime", "los",
    "age_at_admission", "is_female", "first_careunit", "admission_type", "hospital_expire_flag"
]
cohort_df = cohort_df[base_cols]
print(f"Base cohort loaded: {len(cohort_df)} unique ICU stays across {cohort_df['subject_id'].nunique()} patients.")


# ------------------------------------------------------------------------------
# 3. VITAL SIGNS EXTRACTION (icu/chartevents.csv)
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 2: Extracting & Aggregating Vital Signs (icu/chartevents)\n" + "="*70)

# MIMIC-IV MetaVision itemids for key vitals:
# - Heart Rate: 220045
# - MAP (Invasive & Non-invasive): 220052, 220181, 225312
# - SpO2: 220277
# - Respiratory Rate: 220210, 224690
# - Temperature: 223761 (Fahrenheit), 223762 (Celsius)
VITAL_ITEMS = {
    220045: "hr",
    220052: "map",
    220181: "map",
    225312: "map",
    220277: "spo2",
    220210: "resp_rate",
    224690: "resp_rate",
    223761: "temp_f",
    223762: "temp_c"
}

df_chartevents = load_mimic_table("icu", "chartevents", [
    "stay_id", "charttime", "itemid", "valuenum"
])
df_chartevents = df_chartevents[df_chartevents["itemid"].isin(VITAL_ITEMS.keys())].copy()
df_chartevents["valuenum"] = pd.to_numeric(df_chartevents["valuenum"], errors="coerce")
df_chartevents.dropna(subset=["valuenum", "stay_id"], inplace=True)
df_chartevents["stay_id"] = df_chartevents["stay_id"].astype(int)

# Map itemids to clean physiological signal names
df_chartevents["vital_name"] = df_chartevents["itemid"].map(VITAL_ITEMS)

# Normalize Temperature readings to Celsius
is_temp_f = df_chartevents["vital_name"] == "temp_f"
df_chartevents.loc[is_temp_f, "valuenum"] = (df_chartevents.loc[is_temp_f, "valuenum"] - 32.0) * (5.0 / 9.0)
df_chartevents.loc[is_temp_f, "vital_name"] = "temp"
df_chartevents.loc[df_chartevents["vital_name"] == "temp_c", "vital_name"] = "temp"

# Plausibility filtering (removes severe artifacts)
plausible_ranges = {
    "hr": (20, 250),
    "map": (20, 220),
    "spo2": (50, 100),
    "resp_rate": (4, 60),
    "temp": (28, 44)
}
for v_name, (low, high) in plausible_ranges.items():
    mask = (df_chartevents["vital_name"] == v_name) & ((df_chartevents["valuenum"] < low) | (df_chartevents["valuenum"] > high))
    df_chartevents = df_chartevents[~mask]

# Aggregate vital metrics per stay_id (mean, min, max)
vitals_summary = df_chartevents.groupby(["stay_id", "vital_name"])["valuenum"].agg(
    mean="mean", min="min", max="max"
).unstack(level="vital_name")

# Flatten multi-index columns: e.g., hr_mean, hr_min, hr_max
vitals_summary.columns = [f"{vital}_{stat}" for stat, vital in vitals_summary.columns]
vitals_summary.reset_index(inplace=True)
print(f"Extracted vital summary statistics for {len(vitals_summary)} ICU stays.")


# ------------------------------------------------------------------------------
# 4. LAB RESULTS EXTRACTION (hosp/labevents.csv)
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 3: Extracting & Aggregating Core Labs (hosp/labevents)\n" + "="*70)

# Key lab itemids from d_labitems:
# - Lactate: 50813
# - Creatinine: 50912
# - White Blood Cells (WBC): 51301, 51300
# - Platelets: 51265
# - Glucose: 50931, 50809
LAB_ITEMS = {
    50813: "lactate",
    50912: "creatinine",
    51301: "wbc",
    51300: "wbc",
    51265: "platelets",
    50931: "glucose",
    50809: "glucose"
}

df_labevents = load_mimic_table("hosp", "labevents", [
    "hadm_id", "charttime", "itemid", "valuenum"
])
df_labevents = df_labevents[df_labevents["itemid"].isin(LAB_ITEMS.keys())].copy()
df_labevents["valuenum"] = pd.to_numeric(df_labevents["valuenum"], errors="coerce")
df_labevents.dropna(subset=["hadm_id", "valuenum"], inplace=True)
df_labevents["hadm_id"] = df_labevents["hadm_id"].astype(int)
df_labevents["lab_name"] = df_labevents["itemid"].map(LAB_ITEMS)

# Aggregate lab metrics per admission (mean, min, max)
labs_summary = df_labevents.groupby(["hadm_id", "lab_name"])["valuenum"].agg(
    mean="mean", min="min", max="max"
).unstack(level="lab_name")

# Flatten columns: e.g., lactate_mean, lactate_max, creatinine_max, etc.
labs_summary.columns = [f"{lab}_{stat}" for stat, lab in labs_summary.columns]
labs_summary.reset_index(inplace=True)
print(f"Extracted lab summary statistics for {len(labs_summary)} hospital admissions.")


# ------------------------------------------------------------------------------
# 5. TARGET LABEL DEFINITION (Vasopressors, Mortality & Deterioration)
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 4: Defining Target Labels (Vasopressor, Mortality, Deterioration)\n" + "="*70)

# MIMIC-IV MetaVision itemids for Vasopressors & Inotropes in inputevents:
# - Norepinephrine: 221906
# - Epinephrine: 221289
# - Phenylephrine: 221749
# - Vasopressin: 222315
# - Dopamine: 221662
VASOPRESSOR_ITEMIDS = [221906, 221289, 221749, 222315, 221662]

df_inputs = load_mimic_table("icu", "inputevents", ["stay_id", "itemid"])
df_vaso = df_inputs[df_inputs["itemid"].isin(VASOPRESSOR_ITEMIDS)].copy()
vaso_stay_ids = set(df_vaso["stay_id"].dropna().astype(int).unique())

print(f"Identified {len(vaso_stay_ids)} ICU stays with active vasopressor infusions.")


# ------------------------------------------------------------------------------
# 6. MASTER MERGE & FEATURE ENGINEERING
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 5: Assembling Final Cohort DataFrame\n" + "="*70)

# Merge Demographics + Vitals on stay_id
cohort_df = cohort_df.merge(vitals_summary, on="stay_id", how="left")

# Merge Labs on hadm_id
cohort_df = cohort_df.merge(labs_summary, on="hadm_id", how="left")

# Assign Target Labels:
# 1. has_vasopressor: 1 if stay_id in vaso_stay_ids else 0
cohort_df["has_vasopressor"] = cohort_df["stay_id"].isin(vaso_stay_ids).astype(int)

# 2. mortality: hospital_expire_flag from admissions
cohort_df["mortality"] = cohort_df["hospital_expire_flag"].fillna(0).astype(int)

# 3. deterioration_event: Composite target (Vasopressor OR In-Hospital Mortality)
cohort_df["deterioration_event"] = (
    (cohort_df["has_vasopressor"] == 1) | (cohort_df["mortality"] == 1)
).astype(int)

# Drop redundant raw identifier column
cohort_df.drop(columns=["hospital_expire_flag"], inplace=True)

# Set index to stay_id for ML consistency
cohort_df.set_index("stay_id", inplace=True)


# ------------------------------------------------------------------------------
# 7. MISSING VALUE HANDLING STRATEGY (FOR XGBOOST)
# ------------------------------------------------------------------------------
# Note: XGBoost supports missing values (NaN) natively via default split directions.
# However, for baseline modeling or tree models requiring imputation, we can also 
# provide an imputed version using median imputation.
numeric_feature_cols = cohort_df.select_dtypes(include=[np.number]).columns.drop(
    ["subject_id", "hadm_id", "has_vasopressor", "mortality", "deterioration_event"], 
    errors="ignore"
)

# Impute median into a dedicated copy or in-place as preferred
imputed_cohort_df = cohort_df.copy()
for col in numeric_feature_cols:
    median_val = cohort_df[col].median()
    imputed_cohort_df[col] = imputed_cohort_df[col].fillna(median_val)


# ------------------------------------------------------------------------------
# 8. DIAGNOSTICS & DATA EXPLORATION SUMMARY
# ------------------------------------------------------------------------------
print("\n" + "="*70 + "\nSTEP 6: Cohort Inspection & Validation Summaries\n" + "="*70)

print("\n--- A. DataFrame Info ---")
cohort_df.info()

print("\n--- B. Head Preview (First 5 Rows) ---")
display(cohort_df.head()) if "display" in globals() else print(cohort_df.head())

print("\n--- C. Missing Value Summary (Top 10 Features) ---")
missing_series = cohort_df.isnull().sum()
missing_pct = (missing_series / len(cohort_df)) * 100
missing_df = pd.DataFrame({"Missing Count": missing_series, "Missing (%)": missing_pct.round(2)})
print(missing_df[missing_df["Missing Count"] > 0].sort_values("Missing Count", ascending=False).head(10))

print("\n--- D. Target Label Distribution ---")
targets = ["has_vasopressor", "mortality", "deterioration_event"]
for tgt in targets:
    counts = cohort_df[tgt].value_counts()
    prevalence = cohort_df[tgt].mean() * 100
    print(f"Target '{tgt}': {counts.to_dict()} (Prevalence: {prevalence:.2f}%)")

print("\n Cohort building complete. 'cohort_df' is ready for modeling.")
```

```python
# ==============================================================================
# AUTOMATED MIMIC-IV DATA WAREHOUSE & EDA PROFILING PIPELINE
# 1. Ingests all tables from hosp/ and icu/ into SQLite ('mimic_demo.db')
# 2. Builds a flat analytical cohort using SQL
# 3. Generates an interactive HTML EDA report (ydata-profiling / sweetviz)
# ==============================================================================

import os
import glob
import sqlite3
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# Configuration
DB_NAME = "mimic_demo.db"
DIRECTORIES = ["hosp", "icu"]
CHUNK_SIZE = 50_000  # Stream large tables (chartevents, labevents) without memory spikes
REPORT_FILENAME = "mimic_eda_report.html"

# ==============================================================================
# STEP 1: AUTOMATED SQLITE DATA WAREHOUSE SETUP
# ==============================================================================
print("=" * 70)
print(f"STEP 1: Ingesting MIMIC-IV tables into SQLite database ('{DB_NAME}')")
print("=" * 70)

conn = sqlite3.connect(DB_NAME)

# Discover all .csv and .csv.gz files in the specified folders
discovered_files = []
for directory in DIRECTORIES:
    if os.path.exists(directory):
        files = glob.glob(os.path.join(directory, "*.csv*"))
        discovered_files.extend(files)

print(f"Discovered {len(discovered_files)} table files to ingest across {DIRECTORIES}.\n")

for file_path in discovered_files:
    # Derive clean table name: e.g. "icu/chartevents.csv.gz" -> "chartevents"
    base_name = os.path.basename(file_path)
    table_name = base_name.replace(".csv.gz", "").replace(".csv", "")

    print(f"--> Ingesting '{base_name}' into table '{table_name}'...", end=" ", flush=True)

    # Clear existing table if re-running
    conn.execute(f"DROP TABLE IF EXISTS {table_name}")

    # Chunked ingestion for memory efficiency
    total_rows = 0
    for i, chunk in enumerate(pd.read_csv(file_path, chunksize=CHUNK_SIZE, low_memory=False)):
        chunk.to_sql(
            table_name,
            conn,
            if_exists="append",
            index=False,
            method="multi" if len(chunk) < 1000 else None
        )
        total_rows += len(chunk)

    print(f"Done ({total_rows:,} rows).")

# Verify and log all created tables with their SQL row counts
print("\n" + "-" * 50)
print(f"{'Table Name':<25} | {'Row Count':>10}")
print("-" * 50)

cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;")
all_tables = [row[0] for row in cursor.fetchall()]

table_stats = []
for tbl in all_tables:
    count = conn.execute(f"SELECT COUNT(*) FROM {tbl}").fetchone()[0]
    table_stats.append((tbl, count))
    print(f"{tbl:<25} | {count:>10,}")
print("-" * 50)
print(f"Total tables created: {len(table_stats)}\n")

# Speed up queries by indexing key foreign keys
print("Creating indexes on key clinical identifiers...")
for tbl, col in [("icustays", "stay_id"), ("chartevents", "stay_id"), ("inputevents", "stay_id"), ("labevents", "hadm_id")]:
    if tbl in all_tables:
        try:
            conn.execute(f"CREATE INDEX IF NOT EXISTS idx_{tbl}_{col} ON {tbl}({col})")
        except Exception:
            pass
conn.commit()


# ==============================================================================
# STEP 2: BUILD ANALYTICAL COHORT DIRECTLY FROM SQLITE
# ==============================================================================
print("=" * 70)
print("STEP 2: Querying & Assembling Cohort from SQLite")
print("=" * 70)

# 1. Base Cohort & Demographics Query
cohort_query = """
SELECT 
    icu.stay_id,
    icu.subject_id,
    icu.hadm_id,
    icu.first_careunit,
    icu.los,
    p.gender,
    (p.anchor_age + CAST(strftime('%Y', adm.admittime) AS INTEGER) - p.anchor_year) AS age_at_admission,
    adm.admission_type,
    adm.hospital_expire_flag AS mortality
FROM icustays icu
JOIN patients p ON icu.subject_id = p.subject_id
JOIN admissions adm ON icu.subject_id = adm.subject_id AND icu.hadm_id = adm.hadm_id
WHERE icu.los > 0
"""
cohort_df = pd.read_sql_query(cohort_query, conn)
cohort_df = cohort_df[cohort_df["age_at_admission"] >= 18].copy()
cohort_df["is_female"] = (cohort_df["gender"].str.upper() == "F").astype(int)

# 2. Vital Signs Aggregation Query (MetaVision itemids)
vitals_query = """
SELECT 
    stay_id,
    AVG(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_mean,
    MIN(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_min,
    MAX(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_max,
    
    AVG(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_mean,
    MIN(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_min,
    MAX(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_max,
    
    AVG(CASE WHEN itemid = 220277 THEN valuenum END) AS spo2_mean,
    MIN(CASE WHEN itemid = 220277 THEN valuenum END) AS spo2_min,
    
    AVG(CASE WHEN itemid IN (220210, 224690) THEN valuenum END) AS resp_rate_mean,
    MAX(CASE WHEN itemid IN (220210, 224690) THEN valuenum END) AS resp_rate_max,
    
    AVG(CASE 
        WHEN itemid = 223761 THEN (valuenum - 32.0) * (5.0 / 9.0)
        WHEN itemid = 223762 THEN valuenum 
    END) AS temp_mean
FROM chartevents
WHERE itemid IN (220045, 220052, 220181, 225312, 220277, 220210, 224690, 223761, 223762)
  AND valuenum IS NOT NULL
GROUP BY stay_id
"""
df_vitals = pd.read_sql_query(vitals_query, conn)

# 3. Lab Results Aggregation Query
labs_query = """
SELECT 
    hadm_id,
    AVG(CASE WHEN itemid = 50813 THEN valuenum END) AS lactate_mean,
    MAX(CASE WHEN itemid = 50813 THEN valuenum END) AS lactate_max,
    AVG(CASE WHEN itemid = 50912 THEN valuenum END) AS creatinine_mean,
    MAX(CASE WHEN itemid = 50912 THEN valuenum END) AS creatinine_max,
    AVG(CASE WHEN itemid IN (51301, 51300) THEN valuenum END) AS wbc_mean,
    MAX(CASE WHEN itemid IN (51301, 51300) THEN valuenum END) AS wbc_max,
    MIN(CASE WHEN itemid = 51265 THEN valuenum END) AS platelets_min,
    AVG(CASE WHEN itemid IN (50931, 50809) THEN valuenum END) AS glucose_mean
FROM labevents
WHERE itemid IN (50813, 50912, 51301, 51300, 51265, 50931, 50809)
  AND valuenum IS NOT NULL
GROUP BY hadm_id
"""
df_labs = pd.read_sql_query(labs_query, conn)

# 4. Vasopressor Administration Target
vaso_query = """
SELECT DISTINCT stay_id, 1 AS has_vasopressor
FROM inputevents
WHERE itemid IN (221906, 221289, 221749, 222315, 221662)
"""
df_vaso = pd.read_sql_query(vaso_query, conn)

# Merge into unified cohort
cohort_df = cohort_df.merge(df_vitals, on="stay_id", how="left")
cohort_df = cohort_df.merge(df_labs, on="hadm_id", how="left")
cohort_df = cohort_df.merge(df_vaso, on="stay_id", how="left")

# Clean target columns
cohort_df["has_vasopressor"] = cohort_df["has_vasopressor"].fillna(0).astype(int)
cohort_df["mortality"] = cohort_df["mortality"].fillna(0).astype(int)
cohort_df["deterioration_event"] = (
    (cohort_df["has_vasopressor"] == 1) | (cohort_df["mortality"] == 1)
).astype(int)

# Drop database connection
conn.close()

# Prepare final cohort DataFrame for profiling
features_for_profiling = [
    "los", "age_at_admission", "is_female", "first_careunit",
    "hr_mean", "hr_min", "hr_max",
    "map_mean", "map_min", "map_max",
    "spo2_mean", "spo2_min", "resp_rate_mean", "resp_rate_max", "temp_mean",
    "lactate_mean", "lactate_max", "creatinine_mean", "creatinine_max",
    "wbc_mean", "platelets_min", "glucose_mean",
    "has_vasopressor", "mortality", "deterioration_event"
]
profiling_df = cohort_df[features_for_profiling].copy()
print(f"Cohort ready for profiling: {profiling_df.shape[0]} rows, {profiling_df.shape[1]} features.")


# ==============================================================================
# STEP 3: AUTOMATED DATA PROFILING (EDA) REPORT GENERATION
# ==============================================================================
print("\n" + "=" * 70)
print(f"STEP 3: Generating Interactive EDA Report ('{REPORT_FILENAME}')")
print("=" * 70)

report_generated = False

# Try ydata-profiling first
try:
    from ydata_profiling import ProfileReport

    print("Using 'ydata-profiling' to generate comprehensive EDA report...")
    profile = ProfileReport(
        profiling_df,
        title="MIMIC-IV ICU Cohort Exploration & Deterioration Profile",
        dataset={
            "description": "Exploratory Data Analysis for MIMIC-IV Clinical Demo Cohort.",
        },
        explorative=True,
        correlations={"pearson": {"calculate": True}, "spearman": {"calculate": True}},
        minimal=False
    )
    profile.to_file(REPORT_FILENAME)
    print(f" Successfully generated '{REPORT_FILENAME}' via ydata-profiling.")
    report_generated = True

except ImportError:
    print(" 'ydata-profiling' not found. Falling back to 'sweetviz'...")
    try:
        import sweetviz as sv

        print("Generating sweetviz EDA report...")
        report = sv.analyze(profiling_df, target_feat="deterioration_event")
        report.show_html(filepath=REPORT_FILENAME, open_browser=False)
        print(f" Successfully generated '{REPORT_FILENAME}' via sweetviz.")
        report_generated = True
    except ImportError:
        print("\n[!] Neither 'ydata-profiling' nor 'sweetviz' is currently installed.")
        print("To generate the interactive report, run:")
        print("    pip install ydata-profiling")
        print("    # or:")
        print("    pip install sweetviz")

        # Fallback: Save standard tabular statistical summary
        summary_csv = "cohort_statistical_summary.csv"
        profiling_df.describe(include="all").T.to_csv(summary_csv)
        print(f"\nExported baseline statistical summary table to '{summary_csv}'.")

if report_generated:
    print(f"\n All steps complete! Open '{REPORT_FILENAME}' in your browser to inspect distributions, correlations, and missingness.")
```

```python
# ==============================================================================
# PIPELINE: MATERIALIZED ICU FEATURE MART GENERATION
# Role: Database Architect & ML Data Engineer
# Purpose: Pre-aggregate high-cardinality time-series into a static, indexed table
# ==============================================================================

import sqlite3
import time
import pandas as pd

DB_PATH = "mimic_demo.db"

# ------------------------------------------------------------------------------
# 1. DATABASE CONNECTION & PERFORMANCE PRAGMAS
# ------------------------------------------------------------------------------
print("=" * 75)
print(f"Connecting to database '{DB_PATH}' and tuning SQLite engine for bulk write...")
print("=" * 75)

conn = sqlite3.connect(DB_PATH)

# High-performance PRAGMAs for fast analytical transformation & materialization
pragmas = [
    "PRAGMA journal_mode = WAL;",       # Write-Ahead Logging for speed & concurrent reads
    "PRAGMA synchronous = OFF;",        # Disable fsync bottlenecks during table build
    "PRAGMA cache_size = 100000;",      # Allocate ~100MB page cache in RAM
    "PRAGMA temp_store = MEMORY;",      # Store CTE working tables in memory
    "PRAGMA mmap_size = 268435456;"     # Enable 256MB memory-mapped I/O
]
for pragma in pragmas:
    conn.execute(pragma)

# ------------------------------------------------------------------------------
# 2. SQL MATERIALIZATION SCRIPT (CTEs PREVENT CARTESIAN EXPLOSION)
# ------------------------------------------------------------------------------
create_feature_mart_sql = """
-- 1. Idempotency: Drop previous version if exists
DROP TABLE IF EXISTS icu_feature_mart;

-- 2. Materialize the Feature Mart Table
CREATE TABLE icu_feature_mart AS
WITH 
-- CTE 1: Aggregate Vital Signs per stay_id (from icu/chartevents)
vitals_agg AS (
    SELECT 
        stay_id,
        -- Heart Rate (220045)
        AVG(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_mean,
        MIN(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_min,
        MAX(CASE WHEN itemid = 220045 THEN valuenum END) AS hr_max,
        
        -- Systemic MAP: Invasive (220052, 225312) & Non-Invasive (220181)
        AVG(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_mean,
        MIN(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_min,
        MAX(CASE WHEN itemid IN (220052, 220181, 225312) THEN valuenum END) AS map_max,
        
        -- SpO2 (220277)
        AVG(CASE WHEN itemid = 220277 THEN valuenum END) AS spo2_mean,
        MIN(CASE WHEN itemid = 220277 THEN valuenum END) AS spo2_min,
        MAX(CASE WHEN itemid = 220277 THEN valuenum END) AS spo2_max,
        
        -- Respiratory Rate (220210, 224690)
        AVG(CASE WHEN itemid IN (220210, 224690) THEN valuenum END) AS resp_rate_mean,
        MIN(CASE WHEN itemid IN (220210, 224690) THEN valuenum END) AS resp_rate_min,
        MAX(CASE WHEN itemid IN (220210, 224690) THEN valuenum END) AS resp_rate_max,
        
        -- Temperature (°C normalized from 223761 [°F] and 223762 [°C])
        AVG(CASE 
            WHEN itemid = 223761 THEN (valuenum - 32.0) * (5.0 / 9.0)
            WHEN itemid = 223762 THEN valuenum 
        END) AS temp_mean,
        MIN(CASE 
            WHEN itemid = 223761 THEN (valuenum - 32.0) * (5.0 / 9.0)
            WHEN itemid = 223762 THEN valuenum 
        END) AS temp_min,
        MAX(CASE 
            WHEN itemid = 223761 THEN (valuenum - 32.0) * (5.0 / 9.0)
            WHEN itemid = 223762 THEN valuenum 
        END) AS temp_max
    FROM chartevents
    WHERE itemid IN (220045, 220052, 220181, 225312, 220277, 220210, 224690, 223761, 223762)
      AND valuenum IS NOT NULL
    GROUP BY stay_id
),

-- CTE 2: Aggregate Laboratory Results per hadm_id (from hosp/labevents)
labs_agg AS (
    SELECT 
        hadm_id,
        -- Lactate (50813)
        AVG(CASE WHEN itemid = 50813 THEN valuenum END) AS lactate_mean,
        MIN(CASE WHEN itemid = 50813 THEN valuenum END) AS lactate_min,
        MAX(CASE WHEN itemid = 50813 THEN valuenum END) AS lactate_max,
        
        -- Creatinine (50912)
        AVG(CASE WHEN itemid = 50912 THEN valuenum END) AS creatinine_mean,
        MIN(CASE WHEN itemid = 50912 THEN valuenum END) AS creatinine_min,
        MAX(CASE WHEN itemid = 50912 THEN valuenum END) AS creatinine_max,
        
        -- WBC Count (51301, 51300)
        AVG(CASE WHEN itemid IN (51301, 51300) THEN valuenum END) AS wbc_mean,
        MIN(CASE WHEN itemid IN (51301, 51300) THEN valuenum END) AS wbc_min,
        MAX(CASE WHEN itemid IN (51301, 51300) THEN valuenum END) AS wbc_max,
        
        -- Platelets (51265)
        AVG(CASE WHEN itemid = 51265 THEN valuenum END) AS platelets_mean,
        MIN(CASE WHEN itemid = 51265 THEN valuenum END) AS platelets_min,
        MAX(CASE WHEN itemid = 51265 THEN valuenum END) AS platelets_max,
        
        -- Glucose (50931, 50809)
        AVG(CASE WHEN itemid IN (50931, 50809) THEN valuenum END) AS glucose_mean,
        MIN(CASE WHEN itemid IN (50931, 50809) THEN valuenum END) AS glucose_min,
        MAX(CASE WHEN itemid IN (50931, 50809) THEN valuenum END) AS glucose_max
    FROM labevents
    WHERE itemid IN (50813, 50912, 51301, 51300, 51265, 50931, 50809)
      AND valuenum IS NOT NULL
    GROUP BY hadm_id
),

-- CTE 3: Target Event - Vasopressor Support (from icu/inputevents)
vaso_target AS (
    SELECT DISTINCT 
        stay_id, 
        1 AS has_vasopressor
    FROM inputevents
    WHERE itemid IN (221906, 221289, 221749, 222315, 221662)
)

-- Final SELECT: Unified Join to Base Cohort
SELECT 
    icu.stay_id,
    icu.subject_id,
    icu.hadm_id,
    icu.first_careunit,
    icu.los,
    
    -- Exact physiological age at admission: anchor_age + (admittime.year - anchor_year)
    (p.anchor_age + CAST(strftime('%Y', adm.admittime) AS INTEGER) - p.anchor_year) AS age_at_admission,
    CASE WHEN UPPER(p.gender) = 'F' THEN 1 ELSE 0 END AS is_female,
    
    -- Aggregated Vital Signs
    v.hr_mean, v.hr_min, v.hr_max,
    v.map_mean, v.map_min, v.map_max,
    v.spo2_mean, v.spo2_min, v.spo2_max,
    v.resp_rate_mean, v.resp_rate_min, v.resp_rate_max,
    v.temp_mean, v.temp_min, v.temp_max,
    
    -- Aggregated Laboratory Results
    l.lactate_mean, l.lactate_min, l.lactate_max,
    l.creatinine_mean, l.creatinine_min, l.creatinine_max,
    l.wbc_mean, l.wbc_min, l.wbc_max,
    l.platelets_mean, l.platelets_min, l.platelets_max,
    l.glucose_mean, l.glucose_min, l.glucose_max,
    
    -- Ground Truth Target Labels
    COALESCE(vt.has_vasopressor, 0) AS has_vasopressor,
    COALESCE(adm.hospital_expire_flag, 0) AS mortality,
    CASE 
        WHEN COALESCE(vt.has_vasopressor, 0) = 1 OR COALESCE(adm.hospital_expire_flag, 0) = 1 
        THEN 1 
        ELSE 0 
    END AS deterioration_event

FROM icustays icu
INNER JOIN patients p 
    ON icu.subject_id = p.subject_id
INNER JOIN admissions adm 
    ON icu.subject_id = adm.subject_id AND icu.hadm_id = adm.hadm_id
LEFT JOIN vitals_agg v 
    ON icu.stay_id = v.stay_id
LEFT JOIN labs_agg l 
    ON icu.hadm_id = l.hadm_id
LEFT JOIN vaso_target vt 
    ON icu.stay_id = vt.stay_id

-- Filter Adult ICU stays with valid duration
WHERE icu.los > 0
  AND (p.anchor_age + CAST(strftime('%Y', adm.admittime) AS INTEGER) - p.anchor_year) >= 18;
"""

print("Executing materialization query (Aggregating vitals, labs, demographics & targets)...")
t_start = time.perf_counter()
conn.executescript(create_feature_mart_sql)
t_materialize = time.perf_counter() - t_start
print(f" Table 'icu_feature_mart' successfully built in {t_materialize:.3f} seconds.")

# ------------------------------------------------------------------------------
# 3. PERFORMANCE INDEXING
# ------------------------------------------------------------------------------
print("\nCreating indexing on 'icu_feature_mart'...")
conn.execute("CREATE UNIQUE INDEX IF NOT EXISTS idx_mart_stay_id ON icu_feature_mart(stay_id);")
conn.execute("CREATE INDEX IF NOT EXISTS idx_mart_hadm_id ON icu_feature_mart(hadm_id);")
conn.commit()
print(" Unique primary index created on (stay_id).")

# ------------------------------------------------------------------------------
# 4. BENCHMARKING & VALIDATION (SLA < 0.1s)
# ------------------------------------------------------------------------------
print("\n" + "=" * 75)
print("RUNNING LATENCY BENCHMARK: 'SELECT * FROM icu_feature_mart'")
print("=" * 75)

# Benchmark direct zero-join read into Pandas
t_bench_start = time.perf_counter()
mart_df = pd.read_sql_query("SELECT * FROM icu_feature_mart", conn)
t_bench = time.perf_counter() - t_bench_start

print(f" Query Execution Time : {t_bench * 1000:.2f} ms ({t_bench:.4f} seconds)")
if t_bench < 0.10:
    print(f" SLA VERIFIED: Sub-100ms requirement achieved! ({t_bench:.4f}s < 0.1000s)")
else:
    print(f" [!] Notice: Benchmark took {t_bench:.4f}s.")

# Verify Row and Column Counts
print(f"\nDimensions: {mart_df.shape[0]} ICU stays × {mart_df.shape[1]} features")

# Class distribution check
print("\n--- Target Class Balance ---")
for col in ["has_vasopressor", "mortality", "deterioration_event"]:
    pos = mart_df[col].sum()
    pct = (pos / len(mart_df)) * 100
    print(f" - {col:<20}: {pos}/{len(mart_df)} ({pct:.1f}% positive)")

print("\n--- First 3 Records Preview ---")
print(mart_df[["stay_id", "age_at_admission", "los", "hr_mean", "map_mean", "lactate_mean", "deterioration_event"]].head(3))

conn.close()
print("\n 'icu_feature_mart' is ready. All future queries can read from this table without any joins!")
```

```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)

# Connect & query feature mart
conn = sqlite3.connect("mimic_demo.db")
df_mart = pd.read_sql_query("SELECT * FROM icu_feature_mart", conn)
conn.close()

# ------------------------------------------------------------------------------
# Q11: Mean Arterial Pressure (MAP) Hypothesis
# ------------------------------------------------------------------------------
print("=" * 70)
print("Q11: Comparing Mean Arterial Pressure (map_mean & map_min) by Deterioration Status")
print("=" * 70)

map_stats = df_mart.groupby("deterioration_event").agg(
    map_mean_avg=("map_mean", "mean"),
    map_mean_median=("map_mean", "median"),
    map_min_avg=("map_min", "mean"),
    map_min_median=("map_min", "median")
).reset_index()

print(map_stats.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    data=df_mart,
    x="deterioration_event",
    y="map_mean",
    hue="deterioration_event",
    legend=False,
    palette=["#2ecc71", "#e74c3c"],
    ax=axes[0]
)
axes[0].set_title("MAP Mean (mmHg) by Deterioration", fontsize=13, fontweight="bold")
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(["Stable (0)", "Deteriorated (1)"])

sns.boxplot(
    data=df_mart,
    x="deterioration_event",
    y="map_min",
    hue="deterioration_event",
    legend=False,
    palette=["#2ecc71", "#e74c3c"],
    ax=axes[1]
)
axes[1].set_title("MAP Minimum (mmHg) by Deterioration", fontsize=13, fontweight="bold")
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(["Stable (0)", "Deteriorated (1)"])

plt.tight_layout()
plt.show()


# ------------------------------------------------------------------------------
# Q12: Heart Rate vs Vasopressor Support
# ------------------------------------------------------------------------------
print("\n" + "=" * 70)
print("Q12: Comparing Heart Rate (hr_mean & hr_max) between Vasopressor usage")
print("=" * 70)

hr_stats = df_mart.groupby("has_vasopressor").agg(
    hr_mean_avg=("hr_mean", "mean"),
    hr_max_avg=("hr_max", "mean"),
    hr_max_median=("hr_max", "median")
).reset_index()

print(hr_stats.to_string(index=False))

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df_mart,
    x="has_vasopressor",
    y="hr_max",
    hue="has_vasopressor",
    legend=False,
    palette=["#3498db", "#e74c3c"]
)
plt.title("Maximum Heart Rate (hr_max) vs Vasopressor Requirement", fontsize=13, fontweight="bold")
plt.xticks([0, 1], ["No Vasopressor (0)", "Vasopressor Administered (1)"])
plt.xlabel("Vasopressor Status", fontsize=11)
plt.ylabel("Maximum HR (bpm)", fontsize=11)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------------------------
# Q13: Lactate Max & Creatinine Max vs In-Hospital Mortality
# ------------------------------------------------------------------------------
print("\n" + "=" * 70)
print("Q13: Max Lactate & Max Creatinine vs In-Hospital Mortality")
print("=" * 70)

mortality_stats = df_mart.groupby("mortality").agg(
    lactate_max_median=("lactate_max", "median"),
    creatinine_max_median=("creatinine_max", "median")
).reset_index()

print(mortality_stats.to_string(index=False))


# ------------------------------------------------------------------------------
# Q14 & Q15: Top Correlation Heatmap & Length of Stay (LOS)
# ------------------------------------------------------------------------------
print("\n" + "=" * 70)
print("Q14 & Q15: Spearman Correlations with Deterioration Event & LOS Analysis")
print("=" * 70)

# Select numerical features for correlation
num_cols = [
    "age_at_admission", "los", "hr_mean", "hr_max", "map_mean", "map_min",
    "spo2_min", "resp_rate_max", "temp_mean", "lactate_max", "creatinine_max",
    "wbc_max", "platelets_min", "glucose_mean", "deterioration_event"
]

corr_matrix = df_mart[num_cols].corr(method="spearman")
top_corr = corr_matrix[["deterioration_event"]].sort_values("deterioration_event", ascending=False)

print("\n--- Top Spearman Correlations with deterioration_event ---")
print(top_corr)

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix[["deterioration_event"]].sort_values("deterioration_event", ascending=False),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    cbar=True
)
plt.title("Spearman Correlation with Clinical Deterioration", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```
